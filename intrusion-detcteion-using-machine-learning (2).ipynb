# 🔭 Feature Description

<div style="font-family:verdana; font-size: 20px; line-height: 1.7em;"> <ol> <li><p><strong><code>duration</code></strong>: Represents the length of the network connection in seconds. Longer durations may indicate normal sessions, while unusually short or prolonged ones could signal suspicious activity.</p></li> <li><p><strong><code>protocol_type</code></strong>: Specifies the communication protocol used (e.g., TCP, UDP, ICMP). Different protocols exhibit distinct traffic patterns relevant for intrusion classification.</p></li> <li><p><strong><code>service</code></strong>: Denotes the network service accessed (e.g., HTTP, FTP, SSH). Certain services are more vulnerable to specific types of attacks.</p></li> <li><p><strong><code>src_bytes</code></strong>: Number of data bytes sent from the source to the destination. Unusual data transfer volumes may indicate data exfiltration or flooding attacks.</p></li> <li><p><strong><code>dst_bytes</code></strong>: Number of data bytes sent from the destination to the source. Disproportionate values between <code>src_bytes</code> and <code>dst_bytes</code> can signify anomalies.</p></li> <li><p><strong><code>flag</code></strong>: Indicates the status of the network connection based on TCP flags (e.g., SF, REJ). Certain flag combinations are strongly correlated with intrusion attempts.</p></li> <li><p><strong><code>land</code></strong>: A binary feature (1 = same source and destination IP/port, 0 = otherwise). Land attacks exploit this condition to overwhelm target systems.</p></li> <li><p><strong><code>wrong_fragment</code></strong>: Counts the number of incorrect or malformed packets in a connection. High counts often signal packet-based attacks or scanning activity.</p></li> <li><p><strong><code>urgent</code></strong>: Number of urgent packets sent. Excessive urgent packets can indicate malicious behavior like denial-of-service (DoS) attacks.</p></li> <li><p><strong><code>count</code></strong>: Number of connections to the same host in a specified time window. A high value may suggest scanning or probing activity.</p></li> <li><p><strong><code>srv_count</code></strong>: Number of connections to the same service within a time window. This helps identify targeted attacks on specific services.</p></li> <li><p><strong><code>logged_in</code></strong>: A binary indicator (1 = successful login, 0 = failed). Failed logins in large numbers often correspond to brute-force attempts.</p></li> <li><p><strong><code>is_host_login</code></strong>: A binary flag indicating if the connection is a host-level login. Rarely used connections of this type may indicate privilege escalation.</p></li> <li><p><strong><code>is_guest_login</code></strong>: A binary flag (1 = guest login, 0 = otherwise). Guest logins can pose higher security risks due to limited authentication.</p></li> <li><p><strong><code>attack_type</code></strong>: The target variable indicating whether a connection is normal or belongs to a specific intrusion category (e.g., DoS, Probe, R2L, U2R).</p></li> </ol> </div>

# <div style="background-color: #ccf2f4; font-family: verdana; font-size: 20px; line-height: 1.7em; border-radius: 1.3em; padding: 15px;">This code performs the full workflow for a Decision Tree model using scikit-learn 🧠:

Train the Model

Initializes a DecisionTreeClassifier() 🌳

Fits it on the training features and labels (X_train, y_train) ✅

Save the Trained Model

Uses joblib.dump() 💾 to save the trained model as decision_model.pkl for future use

Predict and Evaluate

Makes predictions on the test dataset (X_test) 🔮

Calculates the accuracy of the model using metrics.accuracy_score() 📊

This workflow ensures the model is trained, reusable, and evaluated on unseen data, providing a performance metric for reliability 🤖⚡</div>

```python
import joblib
from sklearn.tree import DecisionTreeClassifier
from sklearn import metrics

# Train Decision Tree Classifier
clf = DecisionTreeClassifier()
clf.fit(X_train, y_train)

# Save the trained model
joblib.dump(clf, "decision_model.pkl")

# Predict and check accuracy
dt_pred = clf.predict(X_test)
print("Accuracy:", metrics.accuracy_score(y_test, dt_pred))
```
